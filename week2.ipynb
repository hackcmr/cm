```python
import numpy as np
def f(x):
    return x**3-x-2
a,b,h=-10,10,1
for x in np.arange(a,b,h):
    if f(x)*f(x+h)<0:
        a=x
        b=x+h
        break
print("The root lies between",a,b)
tol=1e-5
i=1
print(f"\n {"itr":<13}{"a":<10}{"b":<10}{"c":<10}{"f(c)":<10}")
while abs(a-b)/2>tol:
    c=(a+b)/2
    print(f"\n {i:<13}{a:<10.4f}{b:<10.4f}{c:<10.4f}{f(c):<10.4f}")
    if f(a)*f(c)<0:
        b=c
    else:
        a=c
    i=i+1
print(f"\n The required root is {c:.4f}")
```
