```python
import numpy as np
print("1.Power curve\n2.Exponential curve")
c=int(input("Enter the choice:"))
n=int(input("Enter the number of data points:"))
x=np.array(list(map(float,input("Enter the x values").split())))
y=np.array(list(map(float,input("Enter the y values").split())))
 #.........Fitting of a power curve y=ax^b...........
if c==1:
    X=np.log10(x)
    Y=np.log10(y)
    A=np.array([[n,np.sum(X)],[np.sum(X),np.sum(X**2)]])
    B=np.array([np.sum(Y),np.sum(X*Y)])
    log10a,b=np.linalg.solve(A,B)
    a=10*log10a
    print(f"\n The required curve is y={a:.4f} x^{b:.4f}")
#..........Fitting of exponential curve y=ae^(bx)..........
else:
    Y=np.log(y)
    A=np.array([[n,np.sum(x)],[np.sum(x),np.sum(x**2)]])
    B=np.array([np.sum(Y),np.sum(x*Y)])
    loga,b=np.linalg.solve(A,B)
    a=np.exp(loga)
    print(f"\n The required exponential curve is y={a:.4f} e^{b:.4f}x")
```
