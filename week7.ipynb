```python
import math
print("1.Newton's law of cooling 2.Growth 3.Decay")
c=int(input("Enter the choice:"))
if c==1:
    T0,Ts,t1,T1=map(float,input("Enter the values of T0,Ts,t1,T1: ").split())
    k=-math.log((T1-Ts)/(T0-Ts))/t1
    n=int(input("1.Find the temperature 2.FInd the time 3.Both\n choice:"))
    if n in (1,3):
        t=float(input("Enter the given time"))
        print("The required Temperatre=",round(Ts+(T0-Ts)*math.exp(-k*t),2))
    if n in (2,3):
        T=float(input("Enter the given temperature: "))
        print("The required time=",round(-math.log((T-Ts)/(T0-Ts))/k,2),"minutes")
elif c in (2,3):
    N0,t1,N1=map(float,input("Enter the values of N0,t1,N1:").split())
    k=math.log(N1/N0)/t1
    n=int(input("1.Find the Quantity 2.Find the time 3.Both\n choice:"))
    if n in (1,3):
        t=float(input("Enter the given time"))
        print("Growth"if c==2 else "Decay","=",round(N0*math.exp(k*t),2))
    if n in (2,3):
        N=float(input("Enter the given quantity:"))
        print("The required time=",round(math.log(N/N0)/k,2),"hours")
else:
    print("Invalid choice")
```
