```python
import numpy as np
print("1.Straight line\n2.Polynomial")
c=int(input("Enter the choice:"))
n=int(input("Enter the number of data points:"))
x=np.array(list(map(float,input("Enter the x values:").split())))
y=np.array(list(map(float,input("Enter the y values:").split())))
#........fitting a straight line y=a+bx.........
if c==1:
    A=np.array([[n,np.sum(x)],[np.sum(x),np.sum(x**2)]])
    B=np.array([np.sum(y),np.sum(x*y)])
    a,b=np.linalg.solve(A,B)
    print(f"\n The required straight line is y={a:.4f}+{b:.4f}x")
else:
    A=np.array([[n,np.sum(x),np.sum(x**2)],[np.sum(x),np.sum(x**2),np.sum(x**3)],
                [np.sum(x**2),np.sum(x**3),np.sum(x**4)]])
    B=np.array([np.sum(y),np.sum(x*y),np.sum(x**2*y)])
    a,b,c=np.linalg.solve(A,B)
    print(f"\n The required parabola is y={a:.4f}+{b:.4f}x+{c:.4f}x^2")
```
