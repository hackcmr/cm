```python
import numpy as np
import matplotlib.pyplot as plt
n=int(input("Enter the number of data points:"))
x=np.array(list(map(float,input("Enter the x values:").split())))
y=np.array(list(map(float,input("Enter the y values:").split())))
#.........Fit the Straight line y=a+bx............
A=np.array([[n,np.sum(x)],[np.sum(x),np.sum(x**2)]])
B=np.array([np.sum(y),np.sum(x*y)])
a,b=np.linalg.solve(A,B)
print(f"\n The required straight line is y={a:.4f}+{b:.4f}x")
y_fit=a+b*x
plt.scatter(x,y,marker='o',label="Given data")
plt.plot(x,y_fit,label="fitting a straight line")
plt.title("Fitting of straight line")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
n=int(input("Enter the number of data points:"))
x=np.array(list(map(float,input("Enter the x values:").split())))
y=np.array(list(map(float,input("Enter the y values:").split())))
A=np.array([[n,np.sum(x),np.sum(x**2)],[np.sum(x),np.sum(x**2),np.sum(x**3)],[np.sum(x**2),np.sum(x**3),np.sum(x**4)]])
B=np.array([np.sum(y),np.sum(x*y),np.sum(x**2*y)])
a,b,c=np.linalg.solve(A,B)
print(f"\n The required parabola is y={a:.4f}+{b:.4f}x+{c:.4f}x^2")
y_fit=a+b*x+c*x**2
plt.scatter(x,y,marker='o',label="Given data")
plt.plot(x,y_fit,label="fitting a parabola")
plt.title("Fitting of a parabola")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
n=int(input("Enter the number of data points:"))
x=np.array(list(map(float,input("Enter the x values:").split())))
y=np.array(list(map(float,input("Enter the y values:").split())))
X=np.log10(x)
Y=np.log10(y)
A=np.array([[n,np.sum(X)],[np.sum(X),np.sum(X**2)]])
B=np.array([np.sum(Y),np.sum(X*Y)])
log10a,b=np.linalg.solve(A,B)
a=10*log10a
print(f"\n The required curve is y={a:.4f}x^{b:.4f}")
y_fit=a*x**b
plt.scatter(x,y,marker='o',label="Given data")
plt.plot(x,y_fit,label="fitting a power curve")
plt.title("Fitting of power curve")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
n=int(input("Enter the number of data points:"))
x=np.array(list(map(float,input("Enter the x values:").split())))
y=np.array(list(map(float,input("Enter the y values:").split())))
Y=np.log(y)
A=np.array([[n,np.sum(x)],[np.sum(x),np.sum(x**2)]])
B=np.array([np.sum(Y),np.sum(x*Y)])
loga,b=np.linalg.solve(A,B)
a=np.exp(loga)
print(f"\n The required exponential curve is y={a:.4f}e^{b:.4f}x")
y_fit=a*np.exp(b*x)
plt.scatter(x,y,marker='o',label="Given data")
plt.plot(x,y_fit,label="fitting an exponential curve")
plt.title("Fitting of an Exponential curve")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()
plt.show()
```
