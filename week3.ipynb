```python
import sympy as sp
import numpy as np
x=sp.Symbol('x')
f=x**3-x-2
df=sp.diff(f,x)
F=sp.lambdify(x,f)
DF=sp.lambdify(x,df)
a,b,h=-10,10,1
for x in np.arange(a,b,h):
    if F(x)*F(x+h)<0:
        a,b=x,x+h
        break
print("\n The root lies between",a,b)
print("\n The value of the function at end a,b:",F(a),F(b))
tol=1e-4
i=1
print(f"\n{"itr.no":<13}{"x0":<10}{"x1":<10}{"F(x1)"}")
x0=(a+b)/2
while True:
    x1=x0-F(x0)/DF(x0)
    print(f"\n{i:<13}{x0:<10.4f}{x1:<10.4f}{F(x1):.4f}")
    if abs(x1-x0)<tol:
        break
    x0=x1
    i+=1
print(F"\n The required solution is {x1:.4f}")
```
