```python
import sympy as sp
x=sp.symbols('x') # independent variable
y=sp.Function('y') # Dependent variable
n=int(input("Enter the order of given ODE:"))
coeff=[]
for i in range(n+1):
    coeff.append(sp.sympify(input(f"Enter the coefficient of y^{n-i}:")))
lhs=sum(coeff[i]*sp.diff(y(x),x,n-i) for i in range(n+1))
rhs=sp.sympify(input("Enter the RHS term"))
eq=sp.Eq(lhs,rhs)
solution=sp.dsolve(eq)
print("The general solution of given differential is:")
display(solution)
```
