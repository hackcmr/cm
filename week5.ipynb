```python
n=int(input("Enter the no.of equations:"))
a=[]
print("Enter the elements of matrix 'a' row wise")
for i in range(n):
    row=list(map(int,input().split()))
    a.append(row)
print("Enter the elements of matrix 'b'")
b=list(map(int,input().split()))
is_dominant=True
for i in range(n):
    diag=abs(a[i][i])
    row_sum=sum(abs(a[i][j]) for j in range(n) if i!=j)
    if diag<=row_sum:
        is_dominant=False
if is_dominant:
    print("The given system is diagonally dominant")
    x=[0.0]*n
    e=1e-4
    itr=1
    headers="".join(f"x{i+1:<10}" for i in range(n))
    print(f"{"itr.no":<10}{headers}")
    while True:
        x_old=x.copy()
        for i in range(n):
            row_sum1=sum(a[i][j]*x[j] for j in range(n) if i!=j)
            x[i]=(b[i]-row_sum1)/a[i][i]
        values="".join(f"{val:<10.4f}" for val in x)
        print(f"{itr:<10}{values}")
        if all (abs(x[i]-x_old[i])<e for i in range(n)):
            break
        itr=itr+1
    print("\n The final solution is")
    for i in range(n):
        print(f"x{i+1}={x[i]:.4f}") 
else:
    print("The given system is not diagonally dominant")
```
