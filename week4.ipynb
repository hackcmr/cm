```python
# #1 Read the given system in form of matrices
n = int(input("Enter the number of equation: "))
a = []
print("Enter the elements of matrix A, row wise:")
for i in range(n):
    row = list(map(float, input().split()))
    a.append(row)

print("Enter the elements of vector b:")
b = list(map(float, input().split()))

# #2 Check the diagonally dominant condition
is_dominant = True
for i in range(n):
    diag = abs(a[i][i])
    row_sum = sum(abs(a[i][j]) for j in range(n) if i != j)
    if diag <= row_sum:
        is_dominant = False
        break

if is_dominant:
    print("The given system is diagonally dominant")
else:
    print("The given system is not diagonally dominant")

# #3 Jacobi method
x = [0.0] * n
e = 1e-4
itr = 1

headers = "".join(f"x{i+1:<10}" for i in range(n))
print(f"{'itr.no':<10}{headers}")

while True:
    x_new = [0.0] * n
    for i in range(n):
        row_sum = sum(a[i][j] * x[j] for j in range(n) if i != j)
        x_new[i] = (b[i] - row_sum) / a[i][i]

    values = "".join(f"{val:<10.4f}" for val in x_new)
    print(f"{itr:<10}{values}")

    if all(abs(x_new[i] - x[i]) < e for i in range(n)):
        break

    x = x_new.copy()
    itr = itr + 1

print("\nFinal solution is:")
for i in range(n):
    print(f"x{i+1} = {x_new[i]:.4f}")
```
