```python
import math
function=input("Enter the function")
def f(x,y,u):
    return eval(function)
x,y,u,h=map(float,input("Enter the initial values of x,y,u,h=").split())
n=int(input("Enter the number of steps="))
print(f"{'x':<13}{'y':<10}{'u':<10}")
for i in range(n):
    k1=h*u
    l1=h*f(x,y,u)
    k2=h*(u+l1/2)
    l2=h*f(x+h/2,y+k1/2,u+l1/2)
    k3=h*(u+l2/2)
    l3=h*f(x+h/2,y+k2/2,u+l2/2)
    k4=h*(u+l3)
    l4=h*f(x+h,y+k3,u+l3)
    y=y+1/6*(k1+2*k2+2*k3+k4)
    u=u+1/6*(l1+2*l2+2*l3+l4)
    x=x+h
    print(f"{x:5.2f}{y:10.4f}{u:10.4f}")
```
