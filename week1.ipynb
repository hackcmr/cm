```python
import sympy as sp
A=sp.Matrix([[1,4],[3,2]])
eigen_value=A.eigenvects()
for val,mult,vecs in eigen_value:
    print(f"\nThe eigen value is {val} and its A.M is {mult}")
    for v in vecs:
        display("The eigenvector is:",v.T)
```
