```python
import sympy as sp
x,y=sp.symbols('x y')
m=sp.sympify(input("Enter m:"))
n=sp.sympify(input("Enter n:"))
def solve(m,n):
    F=sp.integrate(m,x)
    G=sp.integrate(n-sp.diff(F,y),y)
    return sp.simplify(F+G)
if sp.simplify(sp.diff(m,y)-sp.diff(n,x))==0:
    print("The given equation is exact")
    display('Solution=',solve(m,n),'=c')
else:
    print("The given equation is not exact")
    u=sp.simplify(1/(m*x+n*y))
    if sp.simplify(sp.diff(u*m,y)-sp.diff(u*n,x))!=0:
        u=sp.simplify(1/(m*x-n*y))
    if sp.simplify(sp.diff(u*m,y)-sp.diff(u*n,x))!=0:
        f=sp.simplify((sp.diff(m,y)-sp.diff(n,x))/n)
        if not f.has(y):
            u=sp.exp(sp.integrate(f,x))
    if sp.simplify(sp.diff(u*m,y)-sp.diff(u*n,x))!=0:
        g=sp.simplify((sp.diff(n,x)-sp.diff(m,y))/m)
        if not g.has(x):
            u=sp.exp(sp.integrate(g,y))
    if sp.simplify(sp.diff(u*m,y)-sp.diff(u*n,x))==0:
        display("The integrating factor:",sp.simplify(u))
        display("Solution=",solve(u*m,u*n),"=c")
    else:
        print("Integrating factor is not found")
```
